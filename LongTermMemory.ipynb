```python
from langgraph.store.memory import InMemoryStore
```

```python
store = InMemoryStore()
```

```python
"""" Namespaces is like creating a folder in google drive like (user , u1, profile) -> Inside user folder we have u1 file inside u1 we have profile file"""
namespace = ('user','u1')
```

```python
"""Put method used to create memory it requires(namespace , uniquekey, value/data)"""
store.put(namespace, '1' ,{'data': 'user1 likes momos'})
store.put(namespace, '2', {'data': 'user2 prefer dark mode'})
```

```python
#Adding more Namespace
namespace2 = ('user2','u2')
store.put(namespace2,'1',{'data':'User2 likes Pichy ka Burger'})
store.put(namespace2,'1',{'data': 'User two prefer iphone instead of Android'})
```

```python
"""Retrieving Memory, store.get(namespace, key) """
store.get(namespace,"1")
```

```python
#Retrieving all memories
items = store.search(namespace)
for item in items:
    print(item.value)
```

Long Term Memory if chatbot has 100 memories how it will get to know which memory is relevant to answer the asked question

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

embedding_model = OpenAIEmbeddings(model='text-embedding-3-small')
```

```python
store = InMemoryStore(index={'embed': embedding_model, 'dims':1536}) 
"""means now this store can generate embeddings and return 1536 dimension"""
```

```python
namespace = ('users', 'u1')
```

```python
store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User likes examples in Python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace, "8", {"data": "User is based in India"})
store.put(namespace, "9", {"data": "User likes real-world analogies"})
store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})
```

```python
items = store.search(namespace, query="what is the user currently learning", limit=1)

for item in items:
    print(item.value)
```
